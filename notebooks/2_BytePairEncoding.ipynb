## Load the sequence

```python
with open("../output/combined_text.txt", "r", encoding="utf-8") as f:
    text_sequence = f.read()

len(text_sequence)
```

## BPE algorithm

I am using the [minBPE](https://github.com/karpathy/minbpe) repository to tokenize the sequence of text. <br>
here we tell python the base/parent directory/path where to find the minBPE module

```python
import sys
sys.path.append('..')
```

Start by training the tokenizer on the text sequence that you saved in the previous notebook.

```python
from minbpe import BasicTokenizer

tokenizer = BasicTokenizer()
tokenizer.train(text_sequence, vocab_size=1024)
```

Visualize the vocabulary.

```python
vocab = tokenizer.vocab
vocab
```

**Test the tokenizer.**

Test to encode a sample text.

```python
tokenizer.encode("ndo nimefika honey")
```

Test to decode above token.

```python
tokenizer.decode([623, 110, 336, 102, 585, 901, 121])
```

Add special tokens to the vocabulary. These tokens are going to be used a lot in the fine-tuning step.

```python
max_vocab_id = list(tokenizer.vocab.keys())[-1]
tokenizer.special_tokens = {
    "<|startoftext|>": max_vocab_id + 1,
    "<|separator|>": max_vocab_id + 2,
    "<|endoftext|>": max_vocab_id + 3,
    "<|unk|>": max_vocab_id + 4,
    "<|padding|>": max_vocab_id + 5,
}
```

I have more than 29K tokens for training and validation. This is pretty good, but if you can add more, that would be even better.

```python
len(tokenizer.encode(text_sequence))
```

Save the tokenizer

```python
tokenizer.save(file_prefix="../output/tokenizer/my_tokenizer")
```
